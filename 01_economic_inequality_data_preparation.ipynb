## Notebook 01 — Data Preparation

### Purpose
Generate, prepare, clean, transform, validate, and export a synthetic dataset of 1,000 individuals across 5 regions for comparative economic inequality analysis and interactive visualization.

```python
# ============================================================
# 1. IMPORT REQUIRED LIBRARIES
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Reproducibility
RANDOM_SEED = 42
np.random.seed(RANDOM_SEED)

print("Libraries imported successfully.")
print(f"Random seed set to {RANDOM_SEED}.")
```

```python
# ============================================================
# 2. DEFINE SYNTHETIC DATASET STRUCTURE
# ============================================================

N_INDIVIDUALS = 1000

regions = [
    "North",
    "South",
    "East",
    "West",
    "Central"
]

education_levels = [
    "High School",
    "Diploma",
    "Bachelor",
    "Master",
    "Doctorate"
]

employment_statuses = [
    "Employed",
    "Self-Employed",
    "Unemployed"
]

print(f"Individuals: {N_INDIVIDUALS}")
print(f"Regions: {len(regions)}")
print(f"Education levels: {len(education_levels)}")
print(f"Employment categories: {len(employment_statuses)}")
```

```python
# ============================================================
# 3. GENERATE INDIVIDUAL IDs AND REGIONS
# ============================================================

ids = np.arange(1, N_INDIVIDUALS + 1)

region_data = np.random.choice(
    regions,
    size=N_INDIVIDUALS,
    p=[0.20, 0.20, 0.20, 0.20, 0.20]
)

df = pd.DataFrame({
    "ID": ids,
    "Region": region_data
})

print(df.head())
print("\nIndividuals per region:")
print(df["Region"].value_counts())
```

```python
# ============================================================
# 4. GENERATE EDUCATION LEVEL
# ============================================================

education_probabilities = [
    0.30,  # High School
    0.20,  # Diploma
    0.30,  # Bachelor
    0.15,  # Master
    0.05   # Doctorate
]

df["Education Level"] = np.random.choice(
    education_levels,
    size=N_INDIVIDUALS,
    p=education_probabilities
)

print("Education level distribution:")
print(df["Education Level"].value_counts())
```

```python
# ============================================================
# 5. GENERATE EMPLOYMENT STATUS
# ============================================================

employment_probabilities = {
    "High School": [0.60, 0.15, 0.25],
    "Diploma":     [0.68, 0.17, 0.15],
    "Bachelor":    [0.76, 0.14, 0.10],
    "Master":      [0.82, 0.12, 0.06],
    "Doctorate":   [0.86, 0.10, 0.04]
}

df["Employment Status"] = [
    np.random.choice(
        employment_statuses,
        p=employment_probabilities[education]
    )
    for education in df["Education Level"]
]

print("Employment status distribution:")
print(df["Employment Status"].value_counts())
```

```python
# ============================================================
# 6. GENERATE INCOME
# ============================================================

education_income_multiplier = {
    "High School": 0.75,
    "Diploma": 0.90,
    "Bachelor": 1.15,
    "Master": 1.40,
    "Doctorate": 1.70
}

employment_income_multiplier = {
    "Employed": 1.00,
    "Self-Employed": 1.10,
    "Unemployed": 0.30
}

base_income = np.random.lognormal(
    mean=10.5,
    sigma=0.55,
    size=N_INDIVIDUALS
)

df["Income"] = [
    round(
        base_income[i]
        * education_income_multiplier[df.loc[i, "Education Level"]]
        * employment_income_multiplier[df.loc[i, "Employment Status"]],
        2
    )
    for i in range(N_INDIVIDUALS)
]

print(df["Income"].describe().round(2))
```

```python
# ============================================================
# 7. GENERATE WEALTH
# ============================================================

wealth_multiplier = np.random.lognormal(
    mean=1.5,
    sigma=0.65,
    size=N_INDIVIDUALS
)

df["Wealth"] = (
    df["Income"] * wealth_multiplier
).round(2)

print(df["Wealth"].describe().round(2))
```

```python
# ============================================================
# 8. INSPECT COMPLETE RAW DATASET
# ============================================================

print("Dataset shape:", df.shape)
display(df.head(10))

print("\nData types:")
print(df.dtypes)
```

```python
# ============================================================
# 9. DATA QUALITY AUDIT
# ============================================================

print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate IDs:", df["ID"].duplicated().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nInvalid income values:", (df["Income"] <= 0).sum())
print("Invalid wealth values:", (df["Wealth"] < 0).sum())
```

```python
# ============================================================
# 10. NORMALIZE INCOME AND WEALTH
# ============================================================

df["Income_Normalized"] = (
    (df["Income"] - df["Income"].min()) /
    (df["Income"].max() - df["Income"].min())
)

df["Wealth_Normalized"] = (
    (df["Wealth"] - df["Wealth"].min()) /
    (df["Wealth"].max() - df["Wealth"].min())
)

print(df[[
    "Income", "Income_Normalized",
    "Wealth", "Wealth_Normalized"
]].head())

print("\nNormalized ranges:")
print("Income:", df["Income_Normalized"].min(), "to", df["Income_Normalized"].max())
print("Wealth:", df["Wealth_Normalized"].min(), "to", df["Wealth_Normalized"].max())
```

```python
# ============================================================
# 11. ENCODE CATEGORICAL VARIABLES
# ============================================================

education_encoding = {
    "High School": 1,
    "Diploma": 2,
    "Bachelor": 3,
    "Master": 4,
    "Doctorate": 5
}

employment_encoding = {
    "Unemployed": 0,
    "Self-Employed": 1,
    "Employed": 2
}

region_encoding = {
    "North": 1,
    "South": 2,
    "East": 3,
    "West": 4,
    "Central": 5
}

df["Education_Code"] = df["Education Level"].map(education_encoding)
df["Employment_Code"] = df["Employment Status"].map(employment_encoding)
df["Region_Code"] = df["Region"].map(region_encoding)

print(df[[
    "Region", "Region_Code",
    "Education Level", "Education_Code",
    "Employment Status", "Employment_Code"
]].head(10))
```

```python
# ============================================================
# 12. CREATE INCOME BRACKETS
# ============================================================

df["Income_Bracket"] = pd.qcut(
    df["Income"],
    q=4,
    labels=["Low Income", "Lower-Middle", "Upper-Middle", "High Income"]
)

print("Income bracket distribution:")
print(df["Income_Bracket"].value_counts().sort_index())
```

```python
# ============================================================
# 13. CALCULATE REGIONAL GINI COEFFICIENTS
# ============================================================

def gini_coefficient(values):
    values = np.asarray(values, dtype=float)
    values = np.sort(values)
    
    n = len(values)
    cumulative = np.cumsum(values)
    
    return (n + 1 - 2 * np.sum(cumulative) / cumulative[-1]) / n


regional_gini = (
    df.groupby("Region")["Income"]
    .apply(gini_coefficient)
    .reset_index(name="Gini_Coefficient")
    .sort_values("Gini_Coefficient", ascending=False)
)

print("Regional Income Gini Coefficients:")
display(regional_gini)
```

```python
# ============================================================
# 14. CREATE REGIONAL INEQUALITY SUMMARY
# ============================================================

regional_summary = (
    df.groupby("Region")
    .agg(
        Population=("ID", "count"),
        Mean_Income=("Income", "mean"),
        Median_Income=("Income", "median"),
        Mean_Wealth=("Wealth", "mean"),
        Median_Wealth=("Wealth", "median")
    )
    .reset_index()
)

regional_summary = regional_summary.merge(
    regional_gini,
    on="Region"
)

regional_summary = regional_summary.round(2)

display(regional_summary)
```

```python
# ============================================================
# 15. ANALYZE INCOME BY EDUCATION LEVEL
# ============================================================

education_summary = (
    df.groupby("Education Level", observed=True)
    .agg(
        Population=("ID", "count"),
        Mean_Income=("Income", "mean"),
        Median_Income=("Income", "median"),
        Mean_Wealth=("Wealth", "mean")
    )
    .reindex(education_levels)
    .round(2)
)

display(education_summary)
```

```python
# ============================================================
# 16. ANALYZE EMPLOYMENT BARRIERS BY EDUCATION LEVEL
# ============================================================

employment_by_education = pd.crosstab(
    df["Education Level"],
    df["Employment Status"],
    normalize="index"
).mul(100).reindex(education_levels).round(2)

display(employment_by_education)
```

```python
# ============================================================
# 17. ANALYZE INCOME-TO-EDUCATION DISPARITY BY REGION
# ============================================================

region_education_income = pd.pivot_table(
    df,
    values="Income",
    index="Region",
    columns="Education Level",
    aggfunc="mean",
    observed=True
).reindex(columns=education_levels).round(2)

display(region_education_income)
```

```python
# ============================================================
# 18. QUANTIFY REGIONAL EDUCATION-INCOME DISPARITY
# ============================================================

education_income_gap = region_education_income.copy()

education_income_gap["Income_Gap"] = (
    education_income_gap["Doctorate"] -
    education_income_gap["High School"]
)

education_income_gap["Income_Gap_Ratio"] = (
    education_income_gap["Doctorate"] /
    education_income_gap["High School"]
)

disparity_summary = (
    education_income_gap[["Income_Gap", "Income_Gap_Ratio"]]
    .sort_values("Income_Gap", ascending=False)
    .round(2)
)

display(disparity_summary)
```

```python
# ============================================================
# 19. PREPARE SANKEY FLOW DATA
# ============================================================

education_employment_flow = (
    df.groupby(
        ["Education Level", "Employment Status"],
        observed=True
    )
    .size()
    .reset_index(name="Count")
)

employment_income_flow = (
    df.groupby(
        ["Employment Status", "Income_Bracket"],
        observed=True
    )
    .size()
    .reset_index(name="Count")
)

print("Education → Employment:")
display(education_employment_flow)

print("\nEmployment → Income Bracket:")
display(employment_income_flow)
```

```python
# ============================================================
# 20. FINAL PROCESSED DATA VALIDATION
# ============================================================

print("Final dataset shape:", df.shape)
print("Total missing values:", df.isnull().sum().sum())
print("Duplicate IDs:", df["ID"].duplicated().sum())

print("\nColumns:")
for column in df.columns:
    print("-", column)

print("\nTotal records:", len(df))
```

```python
# ============================================================
# 21. EXPORT PROCESSED MASTER DATASET
# ============================================================

output_file = "data/economic_inequality_processed.csv"

df.to_csv(
    output_file,
    index=False
)

print(f"Dataset exported successfully: {output_file}")
print(f"Rows exported: {len(df)}")
print(f"Columns exported: {len(df.columns)}")
```

```python
# ============================================================
# 22. EXPORT REGIONAL INEQUALITY DATA
# ============================================================

regional_summary.to_csv(
    "data/regional_inequality_summary.csv",
    index=False
)

print("Regional inequality summary exported successfully.")
print("Regions exported:", len(regional_summary))
```

```python
# ============================================================
# 23. EXPORT SANKEY FLOW DATA
# ============================================================

education_employment_flow.to_csv(
    "data/sankey_education_employment.csv",
    index=False
)

employment_income_flow.to_csv(
    "data/sankey_employment_income.csv",
    index=False
)

print("Sankey flow datasets exported successfully.")
print("Education → Employment flows:", len(education_employment_flow))
print("Employment → Income flows:", len(employment_income_flow))
```

```python
# ============================================================
# 24. EXPORT EDUCATION-INCOME DISPARITY DATA
# ============================================================

disparity_export = disparity_summary.reset_index()

disparity_export.to_csv(
    "data/regional_education_income_disparity.csv",
    index=False
)

print("Education-income disparity data exported successfully.")
print("Regions exported:", len(disparity_export))
```

```python
# ============================================================
# 25. EXPORT EMPLOYMENT BARRIER DATA
# ============================================================

employment_barrier_export = employment_by_education.reset_index()

employment_barrier_export.to_csv(
    "employment_barriers_by_education.csv",
    index=False
)

print("Employment barrier data exported successfully.")
print("Education levels exported:", len(employment_barrier_export))
```
